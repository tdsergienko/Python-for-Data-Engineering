# Задание 1

Напишите функцию, которая принимает на вход строку и проверяет, является ли она валидным транспортным номером (1 буква, 3 цифры, 2 буквы, 2–3 цифры). Обратите внимание, что не все буквы кириллического алфавита используются в транспортных номерах.

Если номер валиден, то функция должна возвращать отдельно номер и регион.

Примеры работы программы:
```
car_id = 'А222BС96'
Результат: Номер А222BС валиден. Регион: 96.
```

```
car_id = 'АБ22ВВ193'
Результат: Номер не валиден.
```

````python
VALID_LETTERS_RU = 'АВЕКМНОРСТУХ'
VALID_LETTERS_EN = 'ABEKMHOPCTYX'
VALID_LETTERS = VALID_LETTERS_RU + VALID_LETTERS_EN
PATTERN = re.compile(rf'^([{VALID_LETTERS}])(\d{{3}})([{VALID_LETTERS}]{{2}})(\d{{2,3}})$')
````

````python

def ger_license_plate_region(car_id: str) -> int | None:
    match = PATTERN.match(car_id)
    if match is None:
        return None

    return int(match.group(4))


def solve(car_id: str):
    print(f"car_id = '{car_id}'")

    region = ger_license_plate_region(car_id)
    
    if region is None:
        print('Результат: Номер не валиден')
        return

    print(f'Результат: Номер {car_id} валиден. Регион: {region}')
````

### Тесты

````python
assert ger_license_plate_region("А222BС96") == 96 # en
assert ger_license_plate_region("А222ВС96") == 96 # ru
assert ger_license_plate_region("Р097НН197") == 197 # 3 digits region
assert ger_license_plate_region("АБ22ВВ193") is None
````

### Проверка

````python
solve('А222BС96')
solve('АБ22ВВ193')
````

# Задание 2 (не обязательное)
Напишите функцию, которая будет удалять все последовательные повторы слов из заданной строки при помощи регулярных выражений.

Пример работы программы:
```
some_string = 'Напишите функцию функцию, которая будет будет будет будет удалять все все все все последовательные повторы слов из из из из заданной строки строки при помощи регулярных выражений'

Результат: Напишите функцию, которая будет удалять все последовательные повторы слов из заданной строки при помощи регулярных выражений.
```

````python
def remove_duplicated_words(string: str, case_sensetive=False) -> str:
    flags = 0
    if case_sensetive:
        flags |= re.IGNORECASE
    return re.sub(r'\b(\w+)(\s+\1)+\b', r'\1', string, flags=flags)


def solve(some_string: str) -> None:
    print(f"some_string = {some_string}")
    result = remove_duplicated_words(some_string)
    print(f"Результат: {result}")
````

### Тесты

````python
assert remove_duplicated_words("aaa aaa bbb") == "aaa bbb"
assert remove_duplicated_words("aaa AAA bbb") == "aaa AAA bbb"
assert remove_duplicated_words("aaa AAA bbb", case_sensetive=True) == "aaa bbb"

inp = 'Напишите функцию функцию, которая будет будет будет будет удалять все все все все последовательные повторы слов из из из из заданной строки строки при помощи регулярных выражений.'
out = 'Напишите функцию, которая будет удалять все последовательные повторы слов из заданной строки при помощи регулярных выражений.'
assert remove_duplicated_words(inp) == out
````
