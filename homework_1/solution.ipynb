````python
import os
import json

DATA_DIR = "./data"
PURCHASE_LOG = os.path.join(DATA_DIR, "purchase_log.txt")
VISIT_LOG_CSV = os.path.join(DATA_DIR, "visit_log.csv")
FUNNEL_CSV = os.path.join(DATA_DIR, "funnel.csv")
````

## Задание 1
Переведите содержимое файла purchase_log.txt в словарь purchases вида:

{'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

````python
purchase_log = []
purchases = {}

with open(file=PURCHASE_LOG, mode="r", encoding="utf-8") as file:
    purchase_log = file.readlines()

# пропускаю первую строку, поскольку в ней лежит заголовок:
# {'user_id': 'user_id', 'category': 'category'}
for log in purchase_log[1:]:
    purchase = json.loads(log)
    purchases[purchase["user_id"]] = purchase["category"]
````

### Проверка

Вывожу элементы словаря purchases. Ожидаю увидеть, что первые два элемента будут:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

````python
from itertools import islice

for user_id, category in islice(purchases.items(), 2):
    print(user_id, category)
````

## Задание 2
Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. 
Если покупка была, сам файл visit_log.csv изменять не надо.
Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.
Учтите условия для данных:
- содержимое purchase_log.txt помещается в оперативную память компьютера;
- содержимое visit_log.csv — нет, используйте только построчную обработку этого файла.

Примеры работы программы при выводе первых трёх строк файла funnel.csv:
```csv
user_id,source,category
1840e0b9d4,other,Продукты
4e4f90fcfb,context,Электроника
```

````python
with open(file=VISIT_LOG_CSV, mode="r", encoding="utf-8") as visit_log, \
     open(file=FUNNEL_CSV, mode="w", encoding="utf-8") as funnel:

    # записываю заголовок CSV
    funnel.write("user_id,source,category\n")

    # пропускаю заголовок CSV
    next(visit_log)

    for line in visit_log:
        visit = line.strip()

        user_id, source = visit.split(sep=",")
        category = purchases.get(user_id)
        if category is None:
            continue

        funnel.write(f"{user_id},{source},{category}\n")

````
