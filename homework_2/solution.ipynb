# Задание 2 (не обязательное)
Напишите функцию date_range, которая возвращает список дат за период от start_date до end_date. 

Даты должны вводиться в формате YYYY-MM-DD. 

В случае неверного формата или при start_date > end_date должен возвращаться пустой список.

Примеры работы программы:
```
date_range(‘2022-01-01’, ‘2022-01-03’)
[‘2022-01-01’, ‘2022-01-02’, ‘2022-01-03’]
```

````python
def date_range(start_date: str, end_date: str):
    fmt = "%Y-%m-%d"
    try:
        beg = datetime.strptime(start_date, fmt)
        end = datetime.strptime(end_date, fmt)
    except:
        return []
    
    if beg > end:
        return []

    count = (end - beg).days + 1
    
    return [datetime.strftime(beg + timedelta(days=i), fmt) for i in range(count)]
````

````python
res = date_range('2022.01.01', '2022-01-03') # invald input format
assert res == []

res = date_range('2022-01-01', '2022-01-03')
assert res == ['2022-01-01', '2022-01-02', '2022-01-03']
print(res)

res = date_range('2022-01-03', '2022-01-01')
assert res == []
print(res)

res = date_range('2022-02-30', '2022-02-31')
assert res == []
print(res)          
````
